## Prepare the data

```python
import numpy as np
from matplotlib import pyplot as plt
from IPython import display

import torch
from torch import nn
from torch import optim
from torch.utils.data import Dataset, DataLoader

import torchvision
from torchvision.utils import make_grid

device = torch.device("cuda")
```

```python
class CropCelebA64:

    def __call__(self, pic):
        new_pic = pic.crop((15, 40, 178 - 15, 218 - 30))
        return new_pic

    def __repr__(self):
        return self.__class__.__name__ + '()'
```

```python
# Note that for simplicity we reduce the number of channels to 1
# But if you want to be hardcore feel free to comment .Grayscale :)

train_dataset = torchvision.datasets.CelebA(
    root='celeba',
    split='train',
    transform=torchvision.transforms.Compose([
        CropCelebA64(),
        torchvision.transforms.Resize(64),
        torchvision.transforms.RandomHorizontalFlip(),
        torchvision.transforms.ToTensor(),

        torchvision.transforms.Grayscale()
    ]),
    #download=True,   # in case of problems with downloading from gdrive above
)

validation_dataset = torchvision.datasets.CelebA(
    root='celeba',
    split='valid',
    transform=torchvision.transforms.Compose([
        CropCelebA64(),
        torchvision.transforms.Resize(64),
        torchvision.transforms.ToTensor(),

        torchvision.transforms.Grayscale()
    ]),
)
```

```python
# print some images
samples = torch.stack([train_dataset[i][0] for i in range(32, 48)], dim=0)

plt.figure(figsize=(10, 10))
plt.imshow(make_grid(samples, nrow=4).permute(1, 2, 0))
plt.show()
```

```python
BATCH_SIZE = 64

# you may find useful parameters `num_workers` and `pin_memory`
train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(validation_dataset, batch_size=BATCH_SIZE)
```

# Autoencoder(4 points)

Why to use all this complicated formulaes and regularizations, what is the need for variational inference? To analyze the difference, let's first train just an autoencoder on the data:

```python
dimZ = 100  # Considering face reconstruction task, which size of representation seems reasonable?

# Define the decoder and encoder as networks with one hidden fc-layer
# (that means you will have 2 fc layers in each net)
# Use ReLU for hidden layers' activations
# GlorotUniform initialization for W
# Zero initialization for biases
# It's also convenient to put sigmoid activation on output layer to get nice normalized pics


class Autoencoder(nn.Module):

    def __init__(self, latent_dim):
        super(Autoencoder, self).__init__()
        # Encoder
        self.encoder = nn.Sequential(
            nn.Linear(64*64, 512),
            nn.ReLU(),
            nn.Linear(512, latent_dim)
        )
        # Decoder
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 512),
            nn.ReLU(),
            nn.Linear(512, 64*64),
            nn.Sigmoid()
        )

    def forward(self, x):
        x  = x.view(x.size(0), -1)
        latent_code = self.encoder(x)
        reconstruction = self.decoder(latent_code)
        reconstruction = reconstruction.view(reconstruction.size(0), 1, 64, 64)
        return reconstruction, latent_code
```

```python
from itertools import islice
def show_reconstruction(model, loader):
    images, _ = list(islice(loader, 1))[0]
    reconstruction, latent_code = model(images.to(device))

    reconstruction = reconstruction[:16].cpu()
    plt.imshow(make_grid(reconstruction, nrow=4).permute(1, 2, 0))
```

```python

# Create MSE loss function
criterion = torch.nn.MSELoss()

autoencoder = Autoencoder(dimZ).to(device)

# Use Adam optimizer
optimizer = optim.Adam(autoencoder.parameters(), lr=3e-4)
```

```python

# Train your autoencoder
# Visualize progress in reconstruction and loss decay
n_epoch = 10
tr_loss = 0
train_losses, val_losses = [], []

for epoch in range(n_epoch):
    autoencoder.train()
    for img, _ in train_loader:
        imh = img.to(device)
        optimizer.zero_grad()
        reconstruction, latent_code = autoencoder(imh)
        loss = criterion(reconstruction, imh)
        loss.backward()
        optimizer.step()
        tr_loss += loss.item()
    train_losses.append(loss.item())


    autoencoder.eval()
    val_loss = 0
    with torch.no_grad():
        for img, _ in val_loader:
            img = img.to(device)
            reconstruction, latent_code = autoencoder(img)
            loss = criterion(reconstruction, img)
            val_loss += loss.item()
    val_losses.append(loss.item())


    display.clear_output(wait=True)
    plt.subplot(121)
    plt.plot(np.arange(len(train_losses)), train_losses, label='train')
    plt.plot(np.arange(len(val_losses)), val_losses, label='val')
    plt.legend()
    plt.subplot(122)
    show_reconstruction(autoencoder, val_loader)
    plt.show()
```

```python
# Examine the reconstructions
show_reconstruction(autoencoder, val_loader)
```

# Variational Autoencoder(6 points)

Bayesian approach in deep learning considers everything in terms of distributions. Now our encoder generates not just a vector $z$ but posterior ditribution $q(z|x)$. In our case distribution $q$ is Gaussian distibution $N(\mu, \sigma)$ with parameters $\mu$, $\sigma$. Technically, the first difference is that you need to split bottleneck layer in two. One dense layer will generate vector $\mu$, and another will generate vector $\sigma$. Reparametrization trick should be implemented using the **gaussian_sampler** layer, that generates random vetor $\epsilon$ and returns $\mu+\sigma\epsilon \sim N(\mu, \sigma)$ .

Since our decoder is also a function that generates distribution, we need to do the same splitting for output layer. When testing the model we will look only on mean values, so one of the output will be actual autoencoder output.

In this homework we only ask for implementation of the simplest version of VAE - one $z$ sample per input. You can consider to sample several outputs from one input and average them.

```python
# to compare with conventional AE, keep these hyperparameters
# or change them for the values that you used before
dimZ = 100

# define the network
# you can start from https://github.com/pytorch/examples/blob/master/vae/main.py
# or Theano-based examples here https://github.com/Lasagne/Recipes/blob/master/examples/variational_autoencoder/variational_autoencoder.py
# and here https://github.com/y0ast/Variational-Autoencoder/blob/master/VAE.py
# but remember that this is not your ground truth since the data is not MNIST


class VAE(nn.Module):

    def __init__(self, latent_dim):
        super(VAE, self).__init__()
        self.dim = latent_dim
        self.encoder = nn.Sequential(
            nn.Linear(64*64, 512),
            nn.ReLU(),
            nn.Linear(512, latent_dim*2)
        )
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 512),
            nn.ReLU(),
            nn.Linear(512, 64*64),
            nn.Sigmoid()
        )


    def gaussian_sampler(self, mu, logsigma):
        noise = torch.randn_like(mu)
        return noise * logsigma.exp() + mu

    def forward(self, x):
        x = x.view(x.size(0), -1)
        encoded = self.encoder(x)
        latent_mu, latent_logsigma = encoded[:, :self.dim], encoded[:, self.dim:]
        z = self.gaussian_sampler(latent_mu, latent_logsigma)
        reconstruction_mu = self.decoder(z)
        reconstruction_mu = reconstruction_mu.view(-1, 1, 64, 64)
        reconstruction_logsigma = torch.zeros_like(reconstruction_mu)
        return reconstruction_mu, reconstruction_logsigma, latent_mu, latent_logsigma
```

And the last, but not least! Place in the code where the most of the formulaes goes to - optimization objective. The objective for VAE has it's own name - variational lowerbound. And as for any lowerbound our intention is to maximize it. Here it is (for one sample $z$ per input $x$):

$$\mathcal{L} = -D_{KL}(q_{\phi}(z|x)||p(z)) + \log p_{\theta}(x|z) \to max$$

Your next task is to implement two functions that compute KL-divergence and the second term - log-likelihood of an output. Here is some necessary math for your convenience:

$$D_{KL} = -\frac{1}{2}\sum_{i=1}^{dimZ}(1+log(\sigma_i^2)-\mu_i^2-\sigma_i^2)$$
$$\log p_{\theta}(x|z) = \sum_{i=1}^{dimX}\log p_{\theta}(x_i|z)=\sum_{i=1}^{dimX} \log \Big( \frac{1}{\sigma_i\sqrt{2\pi}}e^{-\frac{(\mu_i-x)^2}{2\sigma_i^2}} \Big)=...$$

Don't forget in the code that you are using $\log\sigma$ as variable. Explain, why not $\sigma$?

```python
def KL_divergence(mu_z, logsigma_z):
    kl = -0.5 * torch.sum(1 + 2*logsigma_z - mu_z.pow(2) - logsigma_z.exp().pow(2), dim=1)
    return kl  # Shape: (batch_size,)

def log_likelihood(x, mu_gen, logsigma_gen):
    x_flat = x.view(x.size(0), -1)  # Flatten input image to (batch_size, 64*64)
    mu_gen_flat = mu_gen.view(mu_gen.size(0), -1)  # Flatten mean of reconstructed output
    logsigma_gen_flat = logsigma_gen.view(logsigma_gen.size(0), -1)
    print(x_flat.shape)
    print(mu_gen_flat.shape)
    print(logsigma_gen.shape)
    recon_loss = -0.5 * torch.sum(torch.log(2*np.pi * logsigma_gen.exp().pow(2)) +(mu_gen - x).pow(2), dim = 1)
    return recon_loss

def loss_vae(x, mu_gen, logsigma_gen, mu_z, logsigma_z):
    recon_loss = -log_likelihood(x, mu_gen, logsigma_gen)  # Negative log-likelihood
    kl_div = KL_divergence(mu_z, logsigma_z)
    total_loss = torch.mean(recon_loss + kl_div)
    return total_loss
```

```python
def show_reconstruction_vae(model, loader):
    images, _ = list(islice(loader, 1))[0]
    mu_gen, _, _, _ = model(images.to(device))
    reconstruction = mu_gen[:16].cpu().clamp(0, 1)

    plt.imshow(make_grid(reconstruction, nrow=4).permute(1, 2, 0))
```

And train the model:

```python
criterion = loss_vae
vae = VAE(dimZ).to(device)
optimizer = optim.Adam(vae.parameters(), lr = 3e-4)
```

```python

# train your variational autoencoder
# visualize progress in reconstruction and loss decay

n_epoch = 10
train_losses, val_losses = [], []

for epoch in range(n_epoch):
    vae.train()
    tr_loss = 0
    for img, _ in train_loader:
        img = img.to(device)
        optimizer.zero_grad()
        mu_gen, logsigma_gen, latent_mu, latent_logsigma = vae(img)
        loss = criterion(img, mu_gen, logsigma_gen, latent_mu, latent_logsigma)
        loss.backward()
        optimizer.step()
        tr_loss += loss.item()
    train_losses.append(tr_loss)


    vae.eval()
    with torch.no_grad():
        for img, _ in val_loader:
            img = img.to(device)
            mu_gen, _, latent_mu, latent_logsigma = vae(img)
            loss = criterion(img, mu_gen, latent_logsigma, latent_mu, latent_logsigma)
            val_loss += loss.item()

    val_losses.append(val_loss)


    display.clear_output(wait=True)
    plt.subplot(121)
    plt.plot(np.arange(len(train_losses)), train_losses, label='train')
    plt.plot(np.arange(len(val_losses)), val_losses, label='val')
    plt.legend()
    plt.subplot(122)
    show_reconstruction_vae(vae, val_loader)
    plt.show()
```
